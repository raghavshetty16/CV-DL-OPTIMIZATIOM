Implement Harris Corner Detector and find the corners
Harris Corner Detection is a method used for corner detection in images.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow


path=r'/content/drive/MyDrive/CVClassLectureDemo/house.tif'
# Read the image in grayscale
image = cv2.imread(path, cv2.IMREAD_GRAYSCALE)  # Replace path with the path to your image
print('Original Image')
cv2_imshow(image)

image = np.float32(image)

print("Image shape:",image.shape)

# Calculate Harris corner detection response
#ksize – Aperture parameter for the Sobel() operator
#k - Harris detector free parameter in the equation.
dst = cv2.cornerHarris(image, blockSize=2, ksize=3, k=0.05)
print(dst.shape)

# Results are marked through the dilated corners
dst = cv2.dilate(dst, None)

# Threshold for an optimal value (adjust as needed)
threshold = 0.01 * dst.max()

# Create a mask for corner regions

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Apply the mask to set corners to green in the original image
image[dst > threshold] = [0, 255, 0]  # Green color: [B, G, R] = [0, 255, 0]

# Display the result
cv2_imshow(image)
```
