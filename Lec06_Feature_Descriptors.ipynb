Histogram of Oriented Gradients HOG is an image feature descripts to describe the image based on the gradient’s directions and magnitudes. At the current time, this project supports calculating the following:

Horizontal and vertical gradients. Gradient magnituge. Gradient direction. Histogram for a given cell.

```python
# Histogram of Oriented Gradients (HOG) of an image using Python's scikit-image library
# This code snippet demonstrates how to compute the HOG features of an image using scikit-image's hog function and visualize the resulting HOG image.
# Adjust the parameters such as orientations, pixels_per_cell, and cells_per_block as needed for your specific requirements.

from skimage.feature import hog
from skimage import exposure
import matplotlib.pyplot as plt
import cv2

# Read the image
path = r'/content/drive/MyDrive/CVClassLectureDemo/house.tif'

# Read the image in grayscale
image = cv2.imread(path, cv2.IMREAD_GRAYSCALE)  # Replace path with the path to your image

# Calculate HOG features
hog_features, hog_image = hog(image, orientations=9, pixels_per_cell=(8, 8),
                              cells_per_block=(2, 2), visualize=True, block_norm='L2-Hys')

# Visualize the HOG image
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6), sharex=True, sharey=True)

ax1.imshow(image, cmap=plt.cm.gray)
ax1.set_title('Input Image')

# Rescale histogram for better visualization
hog_image_rescaled = exposure.rescale_intensity(hog_image, in_range=(0, 10))

ax2.imshow(hog_image_rescaled, cmap=plt.cm.gray)
ax2.set_title('HOG')

plt.show()
```

# Implement SIFT descriptors to identifies key points that are distinctive across an image’s width, height, and most importantly, scale.

```python
# Interest Point Detection

import cv2
from google.colab.patches import cv2_imshow

# Read the image
path = r'/content/drive/MyDrive/CVClassLectureDemo/boat.png'

img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)

# Initialize SIFT detector
sift = cv2.SIFT_create()

# Detect keypoints and compute descriptors
keypoints, descriptors = sift.detectAndCompute(img, None)

# Draw keypoints on the image
image_with_keypoints = cv2.drawKeypoints(img, keypoints, None)

# Display the image with keypoints
print('Image with SIFT Keypoints')
cv2_imshow(image_with_keypoints)
```

# Image Matching and Classification using local descriptors and interest point detectors.

```python
import cv2
from google.colab.patches import cv2_imshow
# This code performs SIFT feature extraction, matches descriptors between two images using a Brute-Force Matcher, applies a ratio test to filter good matches,
# and then draws the matches on a new image for visualization. Adjust the ratio value (0.95 in this case) based on your requirement for good matches.


# Match an input image and its rotated version using sift key points

#path = '/content/drive/MyDrive/Computer_Vision_JI/Standard_images/boat.png'
path = r'/content/drive/MyDrive/CVClassLectureDemo/boat.png'

image1 = cv2.imread(path, cv2.IMREAD_GRAYSCALE)  #

# Get the height and width of the image
height, width = image1.shape[:2]

# Define the rotation angle (in degrees)
angle = 45  # Change this value to the desired rotation angle

# Calculate the rotation matrix
#cv2.getRotationMatrix2D(center, angle, scale)
rotation_matrix = cv2.getRotationMatrix2D((width / 2, height / 2), angle, 1)

# Perform the rotation
image2 = cv2.warpAffine(image1, rotation_matrix, (width, height))

# Display the original and rotated images
print('Original Image')
cv2_imshow(image1)
print('Rotated Image')
cv2_imshow(image2)




# Initialize SIFT detector
sift = cv2.SIFT_create()

# Detect keypoints and compute descriptors for both images
keypoints1, descriptors1 = sift.detectAndCompute(image1, None)
keypoints2, descriptors2 = sift.detectAndCompute(image2, None)

# Initialize a Brute-Force Matcher object
bf = cv2.BFMatcher()

# Match descriptors of the two images
matches = bf.knnMatch(descriptors1, descriptors2, k=2)

# Apply ratio test to get good matches
good_matches = []
for m, n in matches:
    if m.distance < 0.09 * n.distance:
        good_matches.append(m)

# Draw matches on a new image
matched_image = cv2.drawMatches(image1, keypoints1, image2, keypoints2, good_matches, None, flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS)

# Display the matched image
print('Image Matching using SIFT')
cv2_imshow(matched_image)
```
